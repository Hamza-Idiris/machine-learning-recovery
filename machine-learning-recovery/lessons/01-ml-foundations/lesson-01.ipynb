## 1. Core idea

In traditional programming, a developer writes explicit rules that turn inputs into outputs. In Machine Learning, we provide examples and a learning algorithm finds patterns that can help make predictions on new data.

**Example:** Give a model many labeled emails (spam / not spam). It learns patterns and can predict the category of a new email.

### Important vocabulary
- **Dataset:** The collection of examples.
- **Sample / row:** One example in the dataset.
- **Feature:** An input variable used by the model.
- **Label / target:** The answer we want the model to predict (in supervised learning).
- **Model:** The learned pattern or function used to make predictions.


## 2. Three major learning types

| Type | Training information | Example |
|---|---|---|
| Supervised learning | Examples include labels | Predict spam vs. not spam |
| Unsupervised learning | No target labels are provided | Group customers by similar behavior |
| Reinforcement learning | An agent learns from actions and rewards | An agent learning to play a game |

**Check your understanding:** Spam detection with emails already marked as spam or not spam is supervised learning because the examples contain known labels.

## 4. Load a real sample dataset

We will use the built-in Iris dataset. It contains flower measurements and a species label. It is a classic educational dataset, not a dataset you need to download.


```python
iris = load_iris(as_frame=True)
df = iris.frame

df.head()
```

## 5. Identify features and the label

The four measurement columns are features. The `target` column is the label we want to predict.


```python
print('Shape (rows, columns):', df.shape)
print('\nColumn names:')
print(df.columns.tolist())
print('\nFirst five rows:')
display(df.head())
```

```python
feature_columns = iris.feature_names
X = df[feature_columns]
y = df['target']

print('Features (X):')
print(X.head())
print('\nLabel (y):')
print(y.head())
print('\nFeature shape:', X.shape)
print('Label shape:', y.shape)
```

## 6. Inspect data quality and labels

Before training a model, inspect the data. Do not automatically delete missing values; first understand what is missing and why.


```python
print('Data types:')
print(df.dtypes)

print('\nMissing values per column:')
print(df.isna().sum())

print('\nDuplicate rows:', df.duplicated().sum())
```

```python
print('Target counts:')
print(y.value_counts().sort_index())

print('\nTarget names:')
for class_id, class_name in enumerate(iris.target_names):
    print(class_id, '→', class_name)
```
